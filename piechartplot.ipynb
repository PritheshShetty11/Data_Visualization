```python
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
```

```python
plt.pie(runs,labels=labels)
plt.title("Runs of batsmna")
```

```python
colors=['#D16060','#1925C2','#1ED640','#EBDD17']
explode=[0.1,0,0,0]
```

```python
plt.pie(runs,labels=labels,colors=colors,wedgeprops={'edgecolor':'blue','linewidth':2,'linestyle':'--','fill':'True','mouseover':'True'},explode=explode)
plt.title("Runs of batsman")
```
