```python
import matplotlib.pyplot as plt

# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 15, 50, 55, 60, 65, 75, 25, 90]

plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [70, 15, 50, 55, 60, 65, 75, 25, 90]
colors=['red' if score<60 else 'green' for score in exam_scores]
size=[score*2 for score in exam_scores]
plt.scatter(study_hours, exam_scores,s=size,c=colors)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [70, 15, 50, 55, 60, 65, 75, 25, 90]
size=[score*2 for score in exam_scores]
plt.scatter(study_hours, exam_scores,s=size,c=exam_scores,cmap='magma')
plt.colorbar(label='scores')
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [70, 15, 50, 55, 60, 65, 75, 25, 90]
size=[score*2 for score in exam_scores]
plt.scatter(study_hours, exam_scores,s=size,c=exam_scores,cmap='binary')
plt.colorbar(label='scores')
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [70, 15, 50, 55, 60, 65, 75, 25, 90]
size=[score*2 for score in exam_scores]
for i in range(len(study_hours)):
    plt.annotate(f'S{i+1}',(study_hours[i]+0.3,exam_scores[i]))
plt.scatter(study_hours, exam_scores,s=size,c=exam_scores,cmap='binary')
plt.colorbar(label='scores')
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```
